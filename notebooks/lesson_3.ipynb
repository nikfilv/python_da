```python
df = pd.read_csv('../data/raw/aged_15plus_unemployment_rate_percent.csv')
```

```python
# pivot - длиннный -> широкий
# melt - широкий -> длиннный
```

```python
df_melt = pd.melt(
        df, 
        id_vars='country', 
        var_name='year',
        value_name='unemployment_15plus'
)
```

```python
df_melt
```

```python
df_melt.info()
```

```python
df_melt['year'] = pd.to_numeric(df_melt['year'])
```

```python
help(pd.to_numeric)
```

```python
df_melt.info()
```

```python
# 1) Удалить - dropna
# 2) Заполнить - fillna
```

```python
mean_unemployment_15plus_world = df_melt['unemployment_15plus'].mean()
```

```python
mean_unemployment_15plus_world
```

```python
df_2 = df_melt.copy()
```

```python
df_2['unemployment_15plus'] = df_2['unemployment_15plus'].fillna(mean_unemployment_15plus_world)
```

```python
df_2
```
