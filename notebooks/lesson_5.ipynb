```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/raw/healthy_eating_dataset.csv')
```

```python
df
```

```python
df.info()
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('Рапределение калорийности') # Заголовок
sns.histplot(df, x="calories")
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('Рапределение содержания белка') # Заголовок
sns.histplot(df, x="protein_g")
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('Рапределение времени готовки') # Заголовок
sns.histplot(df, x="cook_time_min")
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('Ящик с усами для времени готовки') # Заголовок
sns.boxplot(df, x="cook_time_min")
plt.show() # Показать график
```

```python
df['cook_time_min'].describe()
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('Ящик с усами калорийности в разрезе типов диеты') # Заголовок
sns.boxplot(df, x="calories", hue='diet_type')
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('График рассеяние калорийности и жирности') # Заголовок
sns.scatterplot(df, x="calories", y='fat_g')
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('Медианная калорийность по типу диеты') # Заголовок
sns.barplot(df, x="diet_type", y='calories', errorbar=None, estimator='median')
plt.show() # Показать график
```

```python
df_num = df[[
'calories',            
'protein_g',         
'carbs_g',           
'fat_g',             
'fiber_g',           
'sugar_g',           
'sodium_mg',           
'cholesterol_mg',      
'serving_size_g',      
'prep_time_min',       
'cook_time_min',       
'rating',            
'is_healthy', ]         
]
```
