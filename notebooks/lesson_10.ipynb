```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Простая регрессионная модель.
# 1. Загрузка данных
# 2. Предварительная обработка данных
# 3. Выбор целевой переменной и признаков
# 4. Разделение выборки на обучающую и тестовую
# 5. Обучение модели
# 6. Получение предсказаний и оценка качества
# 7. Анализ коэффициентов модели
```

```python
df = pd.read_csv('../data/raw/insurance.csv')
```

```python
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df_processed = pd.get_dummies(df, columns=['sex', 'smoker', 'region'], drop_first=True)
```

```python
df_processed.info()
```

```python
df['region'].value_counts()
```

```python
y = df_processed['charges']
X = df_processed.drop('charges', axis=1)
```

```python
y
```

```python
X
```

```python
y.shape
```

```python
X.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=5)
```

```python
X_train
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, predictions)
```

```python
mse = mean_squared_error(y_test, predictions)
```

```python
rmse = np.sqrt(mse)
```

```python
r2 = r2_score(y_test, predictions)
```

```python
df['charges'].describe()
```

```python
print(f'mae = {mae:.2f}')
print(f'mse = {mse:.2f}')
print(f'rmse = {rmse:.2f}')
print(f'R^2 = {r2:.2f}')
```
