```python
# Задание:
# 1) Скачать 2 показателя с https://www.gapminder.org/data/
# 2) Предобрабоать данные:
#  - Преоброзовать в длинный формат
#  - Удалить null, если есть
#  - Преоброзовать типы
# 3) Объединить датафреймы через merge
# 4) Вычислить новый столбец используя оба показателя
# 5) Сделать сводную таблицу (pivot_table) по общему df
```

```python
import pandas as pd
```

```python
df_pop = pd.read_csv('../data/raw/pop.csv')
```

```python
df_unemployment_15plus = pd.read_csv('../data/raw/aged_15plus_unemployment_rate_percent.csv')
```

```python
df_pop
```

```python
df_pop = pd.melt(
    df_pop, 
    id_vars='country', 
    var_name='year', 
    value_name='population',
)
```

```python
df_pop
```

```python
df_unemployment_15plus = pd.melt(
    df_unemployment_15plus, 
    id_vars='country', 
    var_name='year', 
    value_name='unemployment_15plus',
)
```

```python
df_pop.info()
```

```python
df_pop = df_pop.dropna()
```

```python
df_pop['year'] = pd.to_numeric(df_pop['year'])
```

```python
df_pop['population'] = pd.to_numeric(
    ((
        df_pop['population'].astype(str)\
        .str.replace('M', 'e6'))\
        .str.replace('k', 'e3'))\
        .str.replace('B', 'e9')\
)
```

```python
df_pop.info()
```

```python
df_unemployment_15plus.info()
```

```python
df_unemployment_15plus = df_unemployment_15plus.dropna()
```

```python
df_unemployment_15plus['year'] = pd.to_numeric(
    df_unemployment_15plus['year']
)
```

```python
df_unemployment_15plus.info()
```

```python
df_merged = pd.merge(
    df_pop, 
    df_unemployment_15plus, 
    how='inner',
    left_on=['country', 'year'],
    right_on=['country', 'year']
)
```

```python
df_merged['pop_unemployment'] = df_merged['population'] * df_merged['unemployment_15plus'] /100
```

```python
df_merged.info()
```

```python
pd.pivot_table(
    df_merged[df_merged['year']>=2020], 
    index='country', 
    values=['pop_unemployment', 'unemployment_15plus'], 
    aggfunc=['mean', 'max'],
).sort_values(('max','unemployment_15plus'))
```
