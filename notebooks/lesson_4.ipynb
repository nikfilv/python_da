```python
import pandas as pd
```

```python
# Работа с нексолькими датафреймами
# Объединение дф через concat
# Соединение дф через merge
# Сводные таблицы pivot_table
```

```python
df_pop = pd.read_csv('../data/raw/pop.csv')
```

```python
df_unemployment_15plus = pd.read_csv('../data/raw/aged_15plus_unemployment_rate_percent.csv')
```

```python
df_pop = pd.melt(
    df_pop, 
    id_vars='country', 
    var_name='year', 
    value_name='population'
)
```

```python
df_unemployment_15plus = pd.melt(df_unemployment_15plus, id_vars='country', var_name='year', value_name='unemployment_15plus')
```

```python
df_pop.head(3)
```

```python
df_pop = df_pop.dropna()
```

```python
df_pop.info()
```

```python
df_pop['year'] = pd.to_numeric(df_pop['year'])
```

```python
df_pop['population'] = pd.to_numeric(
    ((
        df_pop['population'].astype(str)\
        .str.replace('M', 'e6'))\
        .str.replace('k', 'e3'))\
        .str.replace('B', 'e9')\
)
```

```python
df_pop.info()
```

```python
df_unemployment_15plus.info()
```

```python
df_unemployment_15plus = df_unemployment_15plus.dropna()
```

```python
df_unemployment_15plus['year'] = pd.to_numeric(df_unemployment_15plus['year'])
```

```python
df_pop.head(3)
```

```python
df_unemployment_15plus.head(3)
```

```python
df_merged = pd.merge(
    df_pop, 
    df_unemployment_15plus, 
    how='inner',
    left_on=['country', 'year'],
    right_on=['country', 'year']
)
```

```python
df_merged['pop_unemployment'] = df_merged['population'] * df_merged['unemployment_15plus'] /100
```

```python
df_pop_predict = df_pop[df_pop['year']>2023].reset_index().drop(columns=['index'])
```
