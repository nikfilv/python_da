```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/raw/Sleep_health_and_lifestyle_dataset.csv')
```

```python
df
```

```python
df.info()
```

```python
df['Blood Pressure High'] = pd.to_numeric(df['Blood Pressure'].str.split('/', expand=True)[0])
df['Blood Pressure Low'] = pd.to_numeric(df['Blood Pressure'].str.split('/', expand=True)[1])
```

```python
df.info()
```

```python
df['Sleep Disorder'] = df['Sleep Disorder'].fillna('Normal')
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('Рапределение возраста') # Заголовок
sns.histplot(df, x="Age")
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('Рапределение длины сна') # Заголовок
sns.histplot(df, x="Sleep Duration")
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('Рапределение длины сна') # Заголовок
sns.histplot(df, x="Sleep Duration", hue='Gender')
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('Ящик с усами для продолжительности сна') # Заголовок
sns.boxplot(df, x="Sleep Duration")
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('Ящик с усами для продолжительности сна') # Заголовок
sns.boxplot(df, x="Sleep Duration", hue='Gender')
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('График рассеяние продолжительности сна и жирности') # Заголовок
sns.scatterplot(df, x="Sleep Duration", y='Age')
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('График рассеяние продолжительности сна и жирности') # Заголовок
sns.scatterplot(df, x="Sleep Duration", y='Quality of Sleep')
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('График рассеяние продолжительности сна и жирности') # Заголовок
sns.scatterplot(df, x="Stress Level", y='Quality of Sleep')
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('График рассеяние продолжительности сна и жирности') # Заголовок
sns.scatterplot(df, x="Stress Level", y='Age')
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('Медианная длина сна для проблем со сном') # Заголовок
sns.barplot(df, x="Sleep Disorder", y='Sleep Duration', errorbar=None, estimator='median')
plt.show() # Показать график
```

```python
plt.figure(figsize=(10,5)) # Размер рисунка
plt.title('Медианная длина сна по занятости') # Заголовок
sns.barplot(df, y="Occupation", x='Sleep Duration', errorbar=None, estimator='median')
plt.show() # Показать график
```
