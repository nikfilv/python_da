```python
import pandas as pd
import numpy as np
```

```python
df_raw = pd.read_csv('../data/raw/ru_sample.csv')
```

```python
df_raw
```

```python
df_city = df_raw[['city', 'neighborhood']].copy()
```

```python
df_city
```

```python
df_city['city_clean'] = df_city['city'].str.lower()
df_city['neighborhood_clean'] = df_city['neighborhood'].str.lower()
```

```python
df_city[df_city['city']=='г. Москва']
```

```python
bad_words = [
    'г.',
    'с.',
    'сельское поселение',
    'городской округ',
    'город',
    'район',
    'поселок',
]
```

```python
for word in bad_words:
    df_city['city_clean'] = df_city['city_clean'].str.replace(word, '')
    df_city['neighborhood_clean'] = df_city['neighborhood_clean'].str.replace(word, '')
```

```python
df_city[df_city['city']=='г. Москва']
```

```python
df_city['city_clean'] = df_city['city_clean'].str.strip()
df_city['neighborhood_clean'] = df_city['neighborhood_clean'].str.strip()
```

```python
df_city[df_city['city']=='г. Москва']
```

```python
df_city[df_city['city'].str.contains('Москва')]['city'].unique()
```

```python
pat_rus = '[а-я]'
df_city['city_is_rus'] = df_city['city_clean'].str.contains(pat_rus)
df_city['neighborhood_is_rus'] = df_city['neighborhood_clean'].str.contains(pat_rus)
```

```python
df_city
```

```python
def select_city_rus(row):
    if row['city_is_rus']:
        return row['city_clean']
    elif row['neighborhood_is_rus']:
        return row['neighborhood_clean']
    else:
        return np.nan
```

```python
df_city['city_rus'] = df_city.apply(select_city_rus, axis=1)
```
