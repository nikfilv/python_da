```python
df = pd.read_csv('../data/raw/aged_15plus_unemployment_rate_percent.csv')
```

```python
df_melt = pd.melt(
    df, 
    id_vars='country', 
    var_name='year', 
    value_name='unemployment_15plus'
)
```

```python
df_melt.info()
```

```python
df_1 = df_melt.copy()
```

```python
df_1['unemployment_15plus'] = df_1['unemployment_15plus'].fillna(-1)
```

```python
df_1
```

```python
df_1.info()
```

```python
df_melt = df_melt.dropna()
```

```python
df_melt.info()
```

```python
df_melt['year'] = pd.to_numeric(df_melt['year'])
```

```python
df_melt.info()
```

```python
df_melt = df_melt.reset_index(drop=True)
```

```python
df_melt[['country', 'unemployment_15plus']].groupby(by='country').agg('mean')
```

```python
df_grp = df_melt[['country', 'unemployment_15plus']].groupby(by='country').agg(['mean', 'median', 'min', 'max', 'count'])
```
